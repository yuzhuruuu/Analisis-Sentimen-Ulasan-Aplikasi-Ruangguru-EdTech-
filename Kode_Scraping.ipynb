```python
# MURNI HANYA UNTUK SCRAPING DAN SIMPAN DATA
from google_play_scraper import Sort, reviews
import pandas as pd

print("Memulai proses scraping Ruangguru...")
result, continuation_token = reviews(
    'com.ruangguru.livestudents',
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=25000
)

df = pd.DataFrame(result)
df = df[['content', 'score']]
df.rename(columns={'content': 'review', 'score': 'rating'}, inplace=True)

df.dropna(inplace=True)
df.drop_duplicates(subset=['review'], inplace=True)

# SIMPAN SEBAGAI DATASET MENTAH
df.to_csv('dataset_ruangguru_mentah.csv', index=False)
print(f"Scraping selesai! {len(df)} data tersimpan ke dataset_ruangguru_mentah.csv")
```
